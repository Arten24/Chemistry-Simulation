**কেন `7` হলো `2`-এর descendant?** `2→5→7`, মাঝখানে node থাকলেও নিচে পৌঁছানো যায়। **পরীক্ষায়:** child মানে এক edge, descendant মানে এক বা একাধিক edge; ancestor উল্টো দিকে।

### স্লাইড 08 / ৭৩ · [মূল PDF — পৃষ্ঠা 08](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=8)



**Height** = node থেকে তার সবচেয়ে নিচের leaf পর্যন্ত **সর্বোচ্চ edge সংখ্যা**; `height(5)=1` (`5→7`), `height(1)=3` (`1→2→5→7`)। **Depth** = root থেকে node পর্যন্ত edge সংখ্যা; `depth(5)=2` (`1→2→5`), `depth(1)=0`। **Level** এই PDF-এ **depth-এর সমান**, তাই `level(5)=2`, `level(1)=0`। কিছু বইয়ে `level=depth+1`—এ তথ্যও মূল স্লাইডে আছে।

**পরীক্ষার শর্টকাট:** ওপর থেকে গুনলে depth, নিচে সবচেয়ে লম্বা পথে গুনলে height; leaf-এর height `0`। root-এর level এই ক্লাসের স্লাইডে `0`—কোন convention চলছে তা আগে লিখো।

### স্লাইড 09 / ৭৩ · [মূল PDF — পৃষ্ঠা 09](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=9)



এটি **“BINARY TREES”** অংশের আলাদা শিরোনাম। আগের চিত্রে root `1`-এর তিনটি child ছিল—তাই সেটা সাধারণ tree, **binary tree নয়**। এখন থেকে কোনো node-এর তৃতীয় child থাকতে পারবে না।

### স্লাইড 10 / ৭৩ · [মূল PDF — পৃষ্ঠা 10](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=10)



**মূল সংজ্ঞা:** “A binary tree is a tree in which each node has at most 2 child nodes.” চিত্রে `1`-এর বাঁয়ে `2`, ডানে `3`; `2`-এর বাঁয়ে `4`, ডানে `5`। `3,4,5`-এর child `0`—এটিও ঠিক, কারণ *at most* মানে `0/1/2`, ঠিক `2` হওয়া বাধ্যতামূলক নয়। **পরীক্ষায়:** binary tree ≠ binary search tree; এখানে node-এর value ছোট/বড় হওয়ার কোনো শর্ত স্লাইড দেয়নি।

### স্লাইড 11 / ৭৩ · [মূল PDF — পৃষ্ঠা 11](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=11)



**স্লাইডে যত property আছে:** প্রতিটি node-এর সর্বোচ্চ দুই child (**left/right**); level `L`-এ সর্বোচ্চ `2^L` node; depth `d`-তে সর্বোচ্চ `2^d` node; height `h` হলে মোট সর্বোচ্চ `2^(h+1)−1` node; `N≥1` node-এর **minimum possible height** (edge-এ মাপা) `floor(log₂ N)`।

**ধরে দেখি:** root level ০ → `1`; level ১ → `2`; level ২ → `4`, মোট সর্বোচ্চ `1+2+4=7`। চিত্রে `N=5, h=2`; `floor(log₂5)=2`। *সর্বোচ্চ* মানে জায়গা আছে, সব জায়গায় node থাকতেই হবে—এমন নয়।

### স্লাইড 12 / ৭৩ · [মূল PDF — পৃষ্ঠা 12](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=12)



এই এক স্লাইডেই **চার ধরনের tree**: **Full** = প্রতিটি non-leaf-এর ঠিক দুই child (উপর-বাঁয়ের `1,2,3,4,5`); **Complete** = শেষ level বাদে সব level পূর্ণ, শেষ level-এ বাঁদিক থেকে একে একে node (উপর-ডানে `1,2,3,4`); **Perfect** = প্রতিটি level পূর্ণ, সব leaf একই depth-এ (নিচ-বাঁয়ে ৭টি অবস্থান); **Balanced** = স্লাইডের ভাষায় root-এর left/right subtree-র height-এর পার্থক্য সর্বোচ্চ `1` (নিচ-ডানে `1,2,3,4,5`)।

**মিল–অমিল:** perfect ⇒ full এবং complete। উপর-বাঁয়ের ৫-node ছবিটি full হওয়ার পাশাপাশি **complete-ও**; তবে **সব** full tree complete নয়। আলাদা উদাহরণ: `1→(2,3), 3→(4,5)`—শেষ level ডানদিকে শুরু, তাই full কিন্তু complete নয়। স্লাইডের **perfect চিত্রে `4` বারবার লেখা**—ছবি বদলাইনি; এগুলো সাতটি *অবস্থান*, label অনন্য হওয়ার শর্ত নয়। Balanced-এর complexity প্রসঙ্গে নিচে সতর্কতা আছে।

### স্লাইড 13 / ৭৩ · [মূল PDF — পৃষ্ঠা 13](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=13)



এটি **“BALANCED BINARY TREE”** অংশের শিরোনাম। পরের স্লাইডে কেন কম height চাই, তার সূত্র; তার পরেরটিতে search/insert/delete-এর সাথে সম্পর্ক।

### স্লাইড 14 / ৭৩ · [মূল PDF — পৃষ্ঠা 14](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=14)



**স্লাইডের হিসাব একই ক্রমে:** height `h` পর্যন্ত প্রতি level যতটা সম্ভব পূর্ণ হলে `n ≤ 1+2+4+⋯+2^h = 2^(h+1)−1` (geometric series)। তাই `n+1 ≤ 2^(h+1)` ⇒ `log₂(n+1) ≤ h+1` ⇒ `h ≥ log₂(n+1)−1`। স্লাইড শেষে `h ≈ log₂(n)` লিখেছে এবং balanced tree-র height এই minimum-এর কাছে থাকে।

**ইন্টিজার উত্তরে:** height পূর্ণ edge সংখ্যা, তাই `h_min = ceil(log₂(n+1))−1 = floor(log₂ n)` (`n≥1`)। যেমন `n=5` হলে `ceil(log₂6)−1=3−1=2`; `n=7` হলেও `2`; `n=8` হলে `3`। inequality-তে সমতা শুধু প্রত্যেক slot ভরলে—যেমন perfect tree।

### স্লাইড 15 / ৭৩ · [মূল PDF — পৃষ্ঠা 15](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=15)



**স্লাইডে:** root-এর দুই subtree-র height-পার্থক্য সর্বোচ্চ `1`—এটিকে balanced বলছে। Balance থাকলে height আনুমানিক `log₂(n)`-এর কাছাকাছি, আর **height-balanced BST**-এ search, insertion, deletion-এর কাজ `O(log n)` হতে পারে; একদিকে শিকলের মতো হলে height সর্বোচ্চ `n−1` edge (স্লাইডে আনুমানিক “n”)।

**শিক্ষামূলক সতর্কতা, আসল স্লাইড অপরিবর্তিত:** কেবল **root**-এ পার্থক্য ≤১ হলেই পুরো tree-র height `O(log n)` নিশ্চিত হয় না; ভিতরের node-গুলোতেও balance বজায় রাখতে হয় (AVL-এর মতো)। Balance হলেই height একদম সর্বনিম্ন **হতে হবে**—এমনও নয়; যথাযথ দাবি `O(log n)`-এর মধ্যে থাকে। আর binary tree ও BST এক নয়—ordered BST এবং balance রক্ষণাবেক্ষণ করলে উল্লিখিত operation-এর গ্যারান্টি প্রযোজ্য।

### স্লাইড 16 / ৭৩ · [মূল PDF — পৃষ্ঠা 16](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=16)



এটি **“BINARY TREE REPRESENTATION”** অংশের শিরোনাম। একটি tree-র গঠন অপরিবর্তিত রেখেই কেবল কোথায় ও কীভাবে node রাখা হয়, সেটি বদলাবে—স্লাইড ১৭–২০।

### স্লাইড 17 / ৭৩ · [মূল PDF — পৃষ্ঠা 17](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=17)



**স্লাইডে একই tree দুইভাবে:** root `3`, left=`5`, right=`7`; `5`-এর left=`6`, right=`8`। **Linked Node Representation:** প্রতিটি node-তে value আর left/right pointer, leaf-এ child pointer `NULL`। **Array Representation:** index `0` ফাঁকা; `a[1]=3,a[2]=5,a[3]=7,a[4]=6,a[5]=8`।

**বোঝার কৌশল:** চিত্রের edge পাল্টাচ্ছি না; প্রথমটিতে কোন node কাকে point করছে, দ্বিতীয়টিতে index-এর নিয়ম দিয়ে একই সম্পর্ক খুঁজছি। পরের দুই পৃষ্ঠা linked; স্লাইড ২০ array।

### স্লাইড 18 / ৭৩ · [মূল PDF — পৃষ্ঠা 18](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=18)



**স্লাইডের C++ `Node`-এ ঠিক তিনটি field:** integer `value`, `Node* left`, `Node* right`। Constructor `Node(int x)`-এ `this->value=x`, `this->left=NULL`, `this->right=NULL`—নতুন node-র অজানা সন্তানকে শুরুতেই NULL রাখা হয়। ডান পাশের আয়তক্ষেত্রে ওই তিন field-ই আঁকা।

**স্লাইডের কোডটি পড়ার সুবিধায় লাইনভেঙে লিখলাম (মূল ছবি ওপরেই):**
```cpp
class Node {
public:
    int value;
    Node* left;
    Node* right;
    Node(int x) {
        this->value = x;
        this->left = NULL;
        this->right = NULL;
    }
};
```
**কেন pointer?** `left`-এ পরের node-এর ঠিকানা; সন্তান না থাকলে `NULL`—`0` value-র node আর no-child এক জিনিস নয়।

### স্লাইড 19 / ৭৩ · [মূল PDF — পৃষ্ঠা 19](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=19)



**স্লাইডের `main()` উদাহরণ:** `new Node(3)` root; `new Node(5)` বাম child; `new Node(7)` ডান child; `root->left=node1`, `root->right=node2`। চিত্রে `5` ও `7`-এর left/right দুটোই `NULL`।
```cpp
int main() {
    Node* root = new Node(3);
    Node* node1 = new Node(5);
    Node* node2 = new Node(7);
    root->left = node1;
    root->right = node2;
}
```
**খেয়াল করো:** `root->left`-এর মধ্যে `5` সরাসরি লেখা হয়নি; `node1` pointer দেওয়া হয়েছে, যেটি value `5`-এর node-তে পৌঁছায়। এই উদাহরণে এখনও `6,8` যোগ করা হয়নি; স্লাইড ১৭-তে তারা আছে।

### স্লাইড 20 / ৭৩ · [মূল PDF — পৃষ্ঠা 20](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=20)



**স্লাইডের 1-based array:** `a[0]` unused; `a[1]=3,a[2]=5,a[3]=7,a[4]=6,a[5]=8`। index `i`-এর **left=`2i`**, **right=`2i+1`**, **parent=`floor(i/2)`**। যেমন `i=2`-তে `5`; left `a[4]=6`, right `a[5]=8`; index `5`-এর parent index `2`। Child না থাকলে স্লাইড বলেছে সংশ্লিষ্ট জায়গায় `NULL` বা `−1`।

**গুরুত্বপূর্ণ সতর্কতা:** tree-তে ফাঁক থাকলে value বাঁ দিকে গুটিয়ে ফেললে `2i/2i+1` আর সঠিক হবে না; মাঝের slot খালি রেখো। স্লাইডের “`i=1 to number of nodes`” টানা index-ধারণা **এই complete আকারের উদাহরণে** খাটে, arbitrary sparse tree-তে প্রয়োজনীয় array index `n` ছাড়িয়ে যেতে পারে। `−1` যদি বৈধ data হয়, সেই sentinel না ব্যবহার করাই নিরাপদ।

### স্লাইড 21 / ৭৩ · [মূল PDF — পৃষ্ঠা 21](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=21)



এটি **“TREE TRAVERSAL TECHNIQUES”** অংশের শিরোনাম। Traversal মানে প্রত্যেক node একবার নির্দিষ্ট নিয়মে *visit/process* করা; কোনটা আগে visit করব তার ওপর pre/in/post/level ভিন্ন হয়।

### স্লাইড 22 / ৭৩ · [মূল PDF — পৃষ্ঠা 22](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=22)



**স্লাইডের শ্রেণিবিভাগ:** **DFS**-এর তিনটি রূপ—Pre-Order, In-Order, Post-Order; **BFS**-এর রূপ—Level-Order। নোটে লেখা “All tree traversal techniques start at root node.”

**সূক্ষ্ম কিন্তু দরকারি পার্থক্য:** algorithm-কে প্রথমে **root দিয়ে call** করি, কিন্তু visit/print root-ই প্রথম হবে কেবল pre-order ও level-order-এ; in-order-এ বাঁদিক ঘুরে, post-order-এ দুই দিক ঘুরে তারপর root print হয়। DFS পথ ধরে গভীরে, BFS একই level আগে শেষ করে।

### স্লাইড 23 / ৭৩ · [মূল PDF — পৃষ্ঠা 23](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=23)



**Pre-order নিয়ম:** **Parent → Left → Right**। স্লাইডের tree `1`-এর left `2`, right `3`; `2`-এর left `4`, right `5`। প্রথমেই `1` print; তারপর `2`, তার `4,5`; শেষে `3` ⇒ **`1 2 4 5 3`**। “Left” মানে বাঁ সাবট্রির *সব* node আগে, শুধু একটি node নয়।

### স্লাইড 24 / ৭৩ · [মূল PDF — পৃষ্ঠা 24](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=24)



**স্লাইডে C++-এর মূল ক্রম:** `NULL` হলে `return`; তারপর **`cout << node->value`**, তারপর `pre_order_traversal(node->left)`, শেষে `pre_order_traversal(node->right)`। অর্থাৎ parent আগে print।
```cpp
void pre_order_traversal(Node* node) {
    if (node == NULL) { return; }
    cout << node->value << " ";
    pre_order_traversal(node->left);
    pre_order_traversal(node->right);
}
```
**ডান পাশের `main()` একটি *আলাদা* tree বানায়**—স্লাইডে যে initialization/link দেখা যাচ্ছে (তিন order-এর code slide এবং BFS code slide-এও এই tree):
```cpp
Node* root = new Node(3);
Node* node1 = new Node(5);
Node* node2 = new Node(7);
Node* node3 = new Node(4);
root->left = node1;
root->right = node2;
node1->left = node3;
pre_order_traversal(root);
```
এখানে output **`3 5 4 7`**। **স্লাইড ২৫–৩৫-এর tree-র label কিন্তু `1,2,3,4,5`**। এই দুটির output গুলিয়ে ফেলবে না। In/Post/Level code slide-এ নির্মাণ একই, শুধু শেষে যথাক্রমে `in_order_traversal(root)`, `post_order_traversal(root)` অথবা `level_order_traversal(root)` কল করে।

### স্লাইড 25 / ৭৩ · [মূল PDF — পৃষ্ঠা 25](./CSE%20203%20B5%20Tree%20Theory-1-36.pdf#page=25)

**C++ কোডে pre-order থেকে পার্থক্য কেবল print-এর জায়গা:** `NULL→return`, **left recursive call**, তারপর `cout << node->value`, শেষে **right recursive call**।
```cpp
void in_order_traversal(Node* node) {
    if (node == NULL) { return; }
    in_order_traversal(node->left);
    cout << node->value << " ";
    in_order_traversal(node->right);
}
```

### স্লাইড 65 / ৭৩ · [মূল PDF — পৃষ্ঠা 29](./CSE%20203%20B5%20Tree%20Theory-37-73.pdf#page=29)



**স্লাইডের পদ্ধতি (BFS-এর একই structure):** একটি **queue** নাও; root-কে push করো; queue খালি না হওয়া পর্যন্ত front থেকে node বের করো ও process করো; left child থাকলে queue-তে দাও, তারপর right child থাকলে queue-তে দাও। FIFO = আগে ঢুকেছে সে আগে বেরোয়, তাই level অনুযায়ী ক্রম থাকে।

**স্লাইডের C++ function-টি লাইন ধরে (মূল ছবি ওপরেই):**
```cpp
void level_order_traversal(Node* node) {
    queue<Node*> q;
    q.push(node);
    while (!q.empty()) {
        Node* current_node = q.front();
        q.pop();
        cout << current_node->value << " ";
        if (current_node->left != NULL) {
            q.push(current_node->left);
        }
        if (current_node->right != NULL) {
            q.push(current_node->right);
        }
    }
}
```
`q.front()` সামনে থাকা node-কে দেখায়; `q.pop()` সেটাকে সরায়; `cout`-এ সেটি output-এ যায়; left আগে push করলে একই level-এ সেটি আগে বেরোবে। ডান পাশের `main()` **আলাদা** `3→(5,7); 5→(4,∅)` বানায়; তার BFS `3 5 7 4`। **ব্যবহারিক সতর্কতা (স্লাইড বদলানো নয়):** স্লাইডের function-এ শুরুতে NULL guard নেই; খালি tree এলে `q.push(NULL)` দিয়ে পরে dereference হবে—বাস্তব কোডে `if(node==NULL) return;` যোগ করতে হয়।

## অনুশীলন ১ — স্লাইডের প্রশ্ন: recursive calls, returns ও outputs

গাছটি এঁকে রাখো:
```text
       1
     /   \
    2     3
   / \
  4   5
```

**টেবিল পড়া:** `call(k)` = `traversal(Node(k))`-এ ঢোকা; `return to k` = child-এর call শেষে আগের `traversal(Node(k))`-এ ফেরা। স্লাইডের টেবিলে শুধু **৫টি non-NULL call + ৫টি return = ১০ লাইন**; `NULL`-এর জন্য অতিরিক্ত call/return স্লাইডে আলাদা নম্বর পায়নি। Output-এর পরিবর্তন দেখে বুঝবে ঠিক কখন visit/print হয়েছে।

### ১(ক) Pre-order: Parent → Left → Right
| Step | স্লাইডের recursive function call / return | এ লাইন শেষ হওয়া পর্যন্ত output | কেন |
|---:|---|---|---|
| 1 | `pre_order_traversal(Node(1))` | `1` | ১-এ ঢুকেই print |
| 2 | `pre_order_traversal(Node(2))` | `1 2` | root-এর left; ২ আগে print |
| 3 | `pre_order_traversal(Node(4))` | `1 2 4` | ২-এর left; leaf ৪ print |
| 4 | `return to pre_order_traversal(Node(2))` | `1 2 4` | ৪ থেকে ফেরা; ২ পুনরায় print নয় |
| 5 | `pre_order_traversal(Node(5))` | `1 2 4 5` | ২-এর right; ৫ print |
| 6 | `return to pre_order_traversal(Node(2))` | `1 2 4 5` | ৫ শেষ |
| 7 | `return to pre_order_traversal(Node(1))` | `1 2 4 5` | বাঁ subtree শেষ |
| 8 | `pre_order_traversal(Node(3))` | `1 2 4 5 3` | ১-এর right; ৩ print |
| 9 | `return to pre_order_traversal(Node(1))` | `1 2 4 5 3` | ডান subtree শেষ |
| 10 | `return to main()` | `1 2 4 5 3` | সব শেষ |

**উত্তর:** `1 2 4 5 3`।

### ১(খ) In-order: Left → Parent → Right
| Step | recursive call / return | output | কখন print হলো |
|---:|---|---|---|
| 1 | `in_order_traversal(Node(1))` | ফাঁকা | বাঁয়ে যাও |
| 2 | `in_order_traversal(Node(2))` | ফাঁকা | আরও বাঁয়ে |
| 3 | `in_order_traversal(Node(4))` | `4` | ৪-এর left NULL → **৪ print** (স্লাইড ৪২-তে দেখানো) |
| 4 | `return to in_order_traversal(Node(2))` | `4 2` | বাঁ শেষ → **২ print** |
| 5 | `in_order_traversal(Node(5))` | `4 2 5` | ৫-এর left NULL → **৫ print** |
| 6 | `return to in_order_traversal(Node(2))` | `4 2 5` | ২ ইতিমধ্যে print |
| 7 | `return to in_order_traversal(Node(1))` | `4 2 5 1` | বাঁ subtree শেষ → **১ print** |
| 8 | `in_order_traversal(Node(3))` | `4 2 5 1 3` | ৩-এর left NULL → **৩ print** |
| 9 | `return to in_order_traversal(Node(1))` | `4 2 5 1 3` | ডান শেষ |
| 10 | `return to main` | `4 2 5 1 3` | সব শেষ |

**উত্তর:** `4 2 5 1 3`। **খেয়াল:** ৩ নম্বর step-এর call স্লাইড ৪১-এ, একই step-এ ৪-এর visit স্লাইড ৪২-এ; স্লাইড ৪২ কোনো নতুন numbered step নয়।

### ১(গ) Post-order: Left → Right → Parent
| Step | recursive call / return | output | কখন print হলো |
|---:|---|---|---|
| 1 | `post_order_traversal(Node(1))` | ফাঁকা | আগে বাঁ |
| 2 | `post_order_traversal(Node(2))` | ফাঁকা | আগে বাঁ |
| 3 | `post_order_traversal(Node(4))` | `4` | ৪-এর দুই child NULL → **৪ print** (স্লাইড ৫৬) |
| 4 | `return to post_order_traversal(Node(2))` | `4` | ২-এর right বাকি |
| 5 | `post_order_traversal(Node(5))` | `4 5` | ৫-এর দুই child NULL → **৫ print** |
| 6 | `return to post_order_traversal(Node(2))` | `4 5 2` | দুই child শেষ → **২ print** |
| 7 | `return to post_order_traversal(Node(1))` | `4 5 2` | ১-এর right বাকি |
| 8 | `post_order_traversal(Node(3))` | `4 5 2 3` | leaf ৩ print |
| 9 | `return to post_order_traversal(Node(1))` | `4 5 2 3 1` | দুই subtree শেষ → **১ print** |
| 10 | `return to main()` | `4 5 2 3 1` | শেষ |

**উত্তর:** `4 5 2 3 1`। Slide 55-এর Step 3 call ও Slide 56-এর visit একই step-এর দুটি দৃশ্য।

> **Base case নিয়ে প্রশ্ন করলে:** প্রতিটি `f(NULL)` সাথে সাথে return করে; PDF-এর ১০-সারির টেবিল এগুলো দেখায় না। সব NULL call-ও গুনতে চাইলে এই ৫-node binary tree-তে মোট **১১টি call** (`৫টি real node + ৬টি NULL`)—দুটি গণনা-পদ্ধতির পার্থক্য স্পষ্ট করে লিখবে।

## অনুশীলন ২ — স্লাইডের কোডের *অন্য* tree

`3→left 5, right 7; 5→left 4;` অন্য কোনো সন্তান নেই। **সমাধান:** pre=`3 5 4 7` (parent আগে), in=`4 5 3 7` (৫-এর আগে ৪, তারপর ৩, শেষে ৭), post=`4 5 7 3` (root শেষে), level=`3 5 7 4` (root→পরের level→শেষ level)। **কেন আলাদা?** কোডের `main()`-এ root-এর value `3`, ছবির ১০-step trace-এ root-এর value `1`।

## অনুশীলন ৩ — “এটি কি tree?”

ধরা যাক node `A,B,C,D`, edge `A–B,B–C,C–D`। **হ্যাঁ:** সব পৌঁছানো যায়, cycle নেই, `n=4`, edge `3=n−1`। যদি `D–A`-ও যোগ করি? **না:** `A→B→C→D→A` cycle এবং ৪টি edge। যদি শুধু `A–B,C–D` থাকে? **না:** দুই অংশ disconnected, edge মাত্র ২। **মনে রাখবে:** connected ও acyclic দুটোই যাচাই।

## অনুশীলন ৪ — ৭-node terminology (স্লাইড ৬–৮)

`1→(2,3,4); 2→(5,6); 5→7`। **উত্তর:** root=`1`; leaf=`{7,6,3,4}`; `7`-এর ancestors=`{5,2,1}`; `2`-এর descendants=`{5,6,7}`; subtree rooted at `2`=`{2,5,6,7}`; `height(5)=1`, `height(1)=3`, `depth(5)=level(5)=2` (PDF-এ root-এর level 0)। **সমাধানের নিয়ম:** ancestor/descendant-এর জন্য উপরে/নিচে edge ধরো, height-এর জন্য সবচেয়ে দীর্ঘ নিচের পথ।

## অনুশীলন ৫ — node-count ও min-height

`n=10` node হলে minimum height কত? `floor(log₂10)=3` edge; কারণ height 2-এ সর্বোচ্চ `2^(2+1)−1=7` node, কিন্তু height 3-এ সর্বোচ্চ `15`। height `h=3` হলে level 0,1,2,3-এ সর্বোচ্চ `1,2,4,8`; মোট `15`। **ভুল নয়:** `15` maximum; `10` node-এর tree-ও height 3 হতে পারে।

## অনুশীলন ৬ — Full / Complete / Perfect / Balanced

স্লাইড ১২-এর **উপর-বাঁয়ের পাঁচ-node ছবি** (`1→(2,3); 2→(4,5)`) **full, complete, balanced**, কিন্তু **perfect নয়** (শেষ level-এ ৪টি slot-এর কেবল প্রথম দুটো আছে)। **উপর-ডানের চার-node ছবি** (`1→(2,3); 2→4`) **complete**, কিন্তু **full নয়** (`2`-এর এক child)। **নিচ-বাঁয়ের সাত-position ছবি** **perfect, full ও complete**; label `4` একাধিকবার আছে, সেটা মূল স্লাইডেরই চিহ্ন। **Full কিন্তু complete নয়** এমন বাড়তি উদাহরণ: `1→(2,3); 3→(4,5)`; last level ডানদিকে শুরু। Balance যাচাইয়ের সময় tree-র ভেতরের node-গুলোকেও দেখবে।

## অনুশীলন ৭ — array representation-এ index

স্লাইড ২০: `a=[unused,3,5,7,6,8]`। `a[2]=5`-এর **left** `a[2×2]=a[4]=6`, **right** `a[2×2+1]=a[5]=8`। `a[5]=8`-এর **parent** `a[floor(5/2)]=a[2]=5`। `a[3]=7`-এর দুই child index `6,7` অনুপস্থিত: NULL/খালি। **Sparse উদাহরণ:** root ৩-এর left নেই, right ৭ হলে index `1=3`, `2=NULL`, `3=7`—৭-কে index ২-তে সরিও না।

## অনুশীলন ৮ — BFS-এর queue নিজে চালাও

| ধাপ | pop/print | নতুন child push | কাজের পরে queue (সামনে→পেছনে) | output |
|---:|---|---|---|---|
| শুরু | — | root `1` | `[1]` | ফাঁকা |
| ১ | `1` | `2,3` | `[2,3]` | `1` |
| ২ | `2` | `4,5` | `[3,4,5]` | `1 2` |
| ৩ | `3` | — | `[4,5]` | `1 2 3` |
| ৪ | `4` | — | `[5]` | `1 2 3 4` |
| ৫ | `5` | — | `[]` | `1 2 3 4 5` |

**কেন ৩ আগে ৪?** ৩ আগেই queue-তে ঢুকেছিল; FIFO অনুসরণ করছি। নিচের ছোট Python verifier স্লাইডের দুই tree-তেই সব output ও ১০টি call/return যাচাই করে; এটি **বাড়তি চর্চা**, স্লাইডের C++ কোড প্রতিস্থাপন নয়।


````python
# OPTIONAL: extra practice verifier (pure Python; Manim or PDFs are not needed).
from dataclasses import dataclass
from collections import deque

@dataclass
class Node:
    value: int
    left: "Node | None" = None
    right: "Node | None" = None

slide_tree = Node(1, Node(2, Node(4), Node(5)), Node(3))
code_tree = Node(3, Node(5, Node(4)), Node(7))

def recursive_trace(root, order):
    """Return (non-NULL call/return rows, visit order); NULL returns are skipped like the slides."""
    assert order in {"pre", "in", "post"}
    rows, visited = [], []
    def walk(node, caller):
        if node is None:  # PDF's recursive base case
            return
        rows.append(f"{len(rows)+1:2}. call {order}_order_traversal(Node({node.value}))")
        if order == "pre":
            visited.append(node.value)
        walk(node.left, node.value)
        if order == "in":
            visited.append(node.value)
        walk(node.right, node.value)
        if order == "post":
            visited.append(node.value)
        rows.append(f"{len(rows)+1:2}. return to " +
                    (f"{order}_order_traversal(Node({caller}))" if caller is not None else "main"))
    walk(root, None)
    return rows, visited

def level_order(root):
    if root is None:  # safe on an empty tree, unlike the missing guard in slide 65
        return []
    queue, output = deque([root]), []
    while queue:
        node = queue.popleft()
        output.append(node.value)
        if node.left is not None:
            queue.append(node.left)
        if node.right is not None:
            queue.append(node.right)
    return output

for order, expected in [("pre", [1, 2, 4, 5, 3]),
                        ("in", [4, 2, 5, 1, 3]),
                        ("post", [4, 5, 2, 3, 1])]:
    steps, output = recursive_trace(slide_tree, order)
    assert len(steps) == 10 and output == expected
    print(f"\n{order.upper()}  output: {output}\n" + "\n".join(steps))
assert level_order(slide_tree) == [1, 2, 3, 4, 5]
assert [recursive_trace(code_tree, o)[1] for o in ('pre', 'in', 'post')] == [
    [3, 5, 4, 7], [4, 5, 3, 7], [4, 5, 7, 3]]
assert level_order(code_tree) == [3, 5, 7, 4]
print("\nসব উত্তর মিলে গেছে। (স্লাইডে NULL calls ১০ সারির বাইরে রাখা আছে।)")
````
